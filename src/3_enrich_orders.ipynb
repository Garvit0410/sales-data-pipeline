# Step 3: Enriched orders

Adds the customer name and country and the product category and sub category to each order. Profit is rounded to 2 decimals.

```python
import importlib
import os
import sys

# make sure the notebook can find common.py and the step files in this folder
sys.path.insert(0, os.getcwd())

import common
enrich_orders = importlib.import_module("3_enrich_orders")

spark = common.get_spark()
```

```python
df_raw_orders = spark.table(common.table_name(common.RAW_ORDERS))
df_customers = spark.table(common.table_name(common.ENRICHED_CUSTOMERS))
df_products = spark.table(common.table_name(common.ENRICHED_PRODUCTS))

df_orders = enrich_orders.build_enriched_orders(df_raw_orders, df_customers, df_products)
df_orders.limit(10).toPandas()
```

## Checks
Every order should appear exactly once. Orders whose product is not in the product file have no category.

```python
print('raw orders     :', df_raw_orders.count())
print('enriched orders:', df_orders.count())
print('no category    :', df_orders.filter('category IS NULL').count())
```

## Save

```python
common.save_table(df_orders, common.ENRICHED_ORDERS)
```
