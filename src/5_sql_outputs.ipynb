# Step 5: SQL reports

The four profit reports, each written as a SQL query. Each one is shown here and saved as a csv file in `results`.

```python
import importlib
import os
import sys

# make sure the notebook can find common.py and the step files in this folder
sys.path.insert(0, os.getcwd())

import common
reports = importlib.import_module("5_sql_outputs")

spark = common.get_spark()

table = common.table_name(common.PROFIT_AGGREGATE)
```

## Profit by year

```python
df_result = reports.profit_by_year(spark, table)
common.save_csv(df_result, 'profit_by_year')
df_result.toPandas()
```

## Profit by year and product category

```python
df_result = reports.profit_by_year_category(spark, table)
common.save_csv(df_result, 'profit_by_year_category')
df_result.toPandas()
```

## Profit by customer

```python
df_result = reports.profit_by_customer(spark, table)
common.save_csv(df_result, 'profit_by_customer')
df_result.limit(20).toPandas()
```

## Profit by customer and year

```python
df_result = reports.profit_by_customer_year(spark, table)
common.save_csv(df_result, 'profit_by_customer_year')
df_result.limit(20).toPandas()
```
