# Step 2: Enriched customers and products

Cleans the customer names and keeps one row per product id.

```python
import importlib
import os
import sys

# make sure the notebook can find common.py and the step files in this folder
sys.path.insert(0, os.getcwd())

import common
enrich = importlib.import_module("2_enrich_customers_products")

spark = common.get_spark()
```

## Customers
The `customer_name_raw` column keeps the original name next to the cleaned one.

```python
df_raw_customers = spark.table(common.table_name(common.RAW_CUSTOMERS))
df_customers = enrich.enrich_customers(df_raw_customers)
print(df_customers.count(), 'customers')
df_customers.select('customer_id', 'customer_name_raw', 'customer_name', 'country').limit(15).toPandas()
```

## Products
The same product id appears once per state in the file, so it is reduced to one row per id.

```python
df_raw_products = spark.table(common.table_name(common.RAW_PRODUCTS))
df_products = enrich.enrich_products(df_raw_products)
print(df_raw_products.count(), 'rows in the file ->', df_products.count(), 'products')
df_products.limit(10).toPandas()
```

## Save

```python
common.save_table(df_customers, common.ENRICHED_CUSTOMERS)
common.save_table(df_products, common.ENRICHED_PRODUCTS)
```
