# Step 1: Raw tables

Reads the customer (xlsx), product (csv) and order (json) files from the `data` folder and saves them as raw tables. A csv copy of each table goes to the `results` folder.

Databricks does not include the Excel reader, so install it first (it does nothing if it is already installed).

```python
import importlib
import os
import sys

# make sure the notebook can find common.py and the step files in this folder
sys.path.insert(0, os.getcwd())

import common
ingestion = importlib.import_module("1_ingestion")

spark = common.get_spark()
```

## Read the three files

```python
df_customers = ingestion.read_customers(spark, common.CUSTOMERS_FILE)
print(df_customers.count(), 'customers')
df_customers.limit(5).toPandas()
```

```python
df_products = ingestion.read_products(spark, common.PRODUCTS_FILE)
print(df_products.count(), 'products')
df_products.limit(5).toPandas()
```

```python
df_orders = ingestion.read_orders(spark, common.ORDERS_FILE)
print(df_orders.count(), 'orders')
df_orders.limit(5).toPandas()
```

## Save the raw tables and their csv files

```python
spark.sql(f"CREATE DATABASE IF NOT EXISTS {common.DATABASE}")
common.save_table(df_customers, common.RAW_CUSTOMERS)
common.save_table(df_products, common.RAW_PRODUCTS)
common.save_table(df_orders, common.RAW_ORDERS)
print('csv files saved in', common.RESULTS_DIR)
```
