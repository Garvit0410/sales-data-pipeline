# Step 4: Profit aggregate

Total profit by year, product category, product sub category and customer.

```python
import importlib
import os
import sys

# make sure the notebook can find common.py and the step files in this folder
sys.path.insert(0, os.getcwd())

import common
aggregate = importlib.import_module("4_profit_aggregate")

spark = common.get_spark()
```

```python
df_orders = spark.table(common.table_name(common.ENRICHED_ORDERS))
df_profit = aggregate.build_profit_aggregate(df_orders)
print(df_profit.count(), 'rows')
df_profit.orderBy('order_year', 'category', 'sub_category', 'customer_id').limit(15).toPandas()
```

Orders without a matching product are grouped under `Unknown`, so the total stays the same.

```python
from pyspark.sql import functions as F

df_profit.agg(F.round(F.sum('total_profit'), 2).alias('total_profit')).toPandas()
```

## Save

```python
common.save_table(df_profit, common.PROFIT_AGGREGATE)
```
