# Run the unit tests

Runs every test file in this folder with pytest. The `tests` and `src` folders must sit side by side, and the data files must be in the volume.

```python
%pip install pytest openpyxl
```

```python
import sys
import pytest

sys.dont_write_bytecode = True

# forget the modules from an earlier run, so the latest files are used
for name in list(sys.modules):
    if name.startswith(("test_", "conftest", "common")) or name[:2] in ("1_", "2_", "3_", "4_", "5_"):
        del sys.modules[name]

result = pytest.main(["-q", "-p", "no:cacheprovider", "."])
print("all tests passed" if result == 0 else "some tests failed")
```
